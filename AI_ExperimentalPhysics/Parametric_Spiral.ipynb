```python

import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D  # needed for 3D plots

# Parameter
t = np.linspace(0, 8*np.pi, 1000)

# Helix parameters
R  = 1.0    # radius
a  = 0.1    # vertical pitch

# 3D helix
x = R * np.cos(t)
y = R * np.sin(t)
z = a * t

fig = plt.figure(figsize=(6, 6))
ax = fig.add_subplot(111, projection='3d')

ax.plot(x, y, z, color='navy', lw=2)
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('z')
ax.set_title('3D Helix (Spiral)')

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D  # needed for 3D plots

# Parameter
t = np.linspace(0, 8*np.pi, 1000)

######## the widiening spiral
R0 = 0.2   # initial radius
k  = 0.05  # radial growth rate

x = (R0 + k*t) * np.cos(t)
y = (R0 + k*t) * np.sin(t)
z = a * t

fig = plt.figure(figsize=(6, 6))
ax = fig.add_subplot(111, projection='3d')

ax.plot(x, y, z, color='darkred', lw=2)
ax.set_title('3D Widening Spiral')

plt.tight_layout()
plt.show()
```
