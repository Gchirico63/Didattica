```python
import numpy as np
import matplotlib.pyplot as plt

xmax = 10
xmin = -xmax
dist = xmax-xmin
Np = 500
k0 = np.pi/dist
xspace = np.linspace(xmin,xmax,Np)

Nmodes = 40
Amplitude = 0.1
Amp = np.random.rand(Nmodes)
Amp = (2*Amp-1)*Amplitude
print (Amp)

Signal = np.zeros(Np)
for im,aa in enumerate(Amp):
  Signal = Signal + aa*np.sin(xspace*k0*im)

F, A = plt.subplots(1,2,figsize=(8,4))
A[0].plot(xspace,Signal)
A[0].text(xmin,Signal.max(), 'N of points='+str(Np))
A[0].set_xlabel('x [m]')
A[0].set_ylabel('y [m]')
A[1].bar(np.arange(1, Nmodes + 1), Amp)
A[1].text(0,Amp.max(), 'N of modes ='+str(Nmodes))
A[1].yaxis.tick_right()
A[1].yaxis.set_label_position("right")
A[1].set_ylabel("modes amplitude")
```

```python
from IPython.display import Audio
import numpy as np
'''
imagine now that instead of a shape of the string it is a sound wave,
so the signal is a signal of time, instead of space.
This allows us to visualize or audialize the signal sending it to a speaker.
We repeat the signal many times to have something audible, that last a few seconds
'''
# Repeat Signal 100 times (total = 10,000 samples) at 44kHz means about 0.25 s
repeated_signal = np.tile(Signal, 200)

# Normalize to [-1, 1]
repeated_signal = repeated_signal / np.max(np.abs(repeated_signal))

# Play at 44.1 kHz (~0.23 seconds at 441 Hz fundamental)
Audio(repeated_signal, rate=44100, autoplay=True)
```
