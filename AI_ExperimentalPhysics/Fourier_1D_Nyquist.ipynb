```python
import numpy as np
import matplotlib.pyplot as plt


tempo = np.linspace(-40,40,100)

dt = tempo[1]-tempo[0]

y = 0.1*np.cos(2*np.pi*0.2*tempo) + 0.4*np.sin(2*np.pi*0.5*tempo)+ 0.3*np.cos(2*np.pi*0.3*tempo)

plt.figure()
plt.plot(tempo,y)
plt.xlabel('time [s]',fontsize= 20)
plt.ylabel('y',fontsize= 20)

dtempo = tempo[1]-tempo[0]

yft = np.fft.fft(y)
xf =np.fft.fftfreq(y.size,dtempo)

plt.figure()
plt.plot(np.fft.fftshift(xf),np.abs(np.fft.fftshift(yft)))
plt.xlabel('freq [Hz]',fontsize= 20)
plt.ylabel('FT-spectrum',fontsize= 20)

```
